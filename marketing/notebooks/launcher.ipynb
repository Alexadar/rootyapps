## Available Styles

- `style_default` - 2.5x zoom, full screenshot
- `style_zoom_high` - 3.0x zoom, full screenshot
- `style_top_zoom` - 2.5x zoom, top 50%
- `style_middle_zoom` - 2.5x zoom, middle 50%
- `style_bottom_zoom` - 2.5x zoom, bottom 50%

## Crop Options

- `'full'` - No crop
- `'top50'` - Top 50%
- `'middle50'` - Middle 50%
- `'bottom50'` - Bottom 50%
- `'top33'` - Top 33%
- `'middle33'` - Middle 33%
- `'bottom33'` - Bottom 33%
